```python
#Import Library
import cv2

print(cv2.__version__)
print(hasattr(cv2, "CascadeClassifier"))
```

```python
# Single Object Detection
# Import Model
front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")

#import Image/Video
image = cv2.imread("vijay.jpg")

# Convert it to GreyScale
grey_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

#using multiscale detector to get the position coordinates
face_detector  = front_face_detector.detectMultiScale(grey_image)
print(face_detector)

#using for loops to draw the rectangular bounding box as much objects we get
for x,y,w,h in face_detector:
    cv2.rectangle(grey_image,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow("image_window",grey_image)

#Waitkey
cv2.waitKey()
```

```python
# Multiple Object Detection

front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")
image = cv2.imread("police_multiple_images.jpg")
grey_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
face_detector  = front_face_detector.detectMultiScale(grey_image)
print(face_detector)

for x,y,w,h in face_detector:
    cv2.rectangle(grey_image,(x,y),(x+w,y+h),(255,0,0),3)
cv2.imshow("image_window",grey_image)
cv2.waitKey()
```

# 3. CCTV/WebCam Detection

```python
front_face_detector = cv2.CascadeClassifier("haarcascade_frontalface_alt.xml")

video = cv2.VideoCapture(0)


while True:
    success,frame = video.read()
    if success == True:
        grey_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(grey_image)

        for x,y,w,h in face_detector:
            cv2.rectangle(grey_image,(x,y),(x+w,y+h),(255,0,0),3)

        cv2.imshow("image_window",grey_image)
        key = cv2.waitKey(1)

        if key == 81 or key == 113:
            break
video.release()
cv2.destroyAllWindows()
```
