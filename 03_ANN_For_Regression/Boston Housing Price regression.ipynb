# Problem Statement

### Build a ML Model that predicts the median house price based on 13 socio-economic and property-related features from the Boston Housing dataset.

## 1.Import Necessary libraries

```python
import numpy as np
import keras
import tensorflow as tf
from keras.datasets import boston_housing
from keras.models import Sequential
from keras.layers import Dense
from sklearn.preprocessing import StandardScaler
from keras.models import load_model
```

## 2.Import Dataset

```python
(X_train,y_train),(X_test,y_test) = boston_housing.load_data()
```

## 3.Data Understanding

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

## 4.Data Preparation

```python
scaler = StandardScaler()

X_train=scaler.fit_transform(X_train)
X_test=scaler.fit_transform(X_test)
```

```python
X_train.mean().round(),X_train.std().round()
```

```python
X_test.mean().round(),X_test.std().round()
```

## 5.Model Building

```python
model = Sequential()

model.add(Dense(64,activation='relu',input_shape=(13,))) # For Structured data, use input_shape
model.add(Dense(32,activation='relu'))
model.add(Dense(16,activation='relu'))

#Output layer
model.add(Dense(1,))  #No activation function -> Becoz, no regression
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

## 6.Model Training | 7.Model Testing | 8.Model Evaluation

```python
model.fit(X_train,y_train,epochs=100,batch_size=16,validation_split=0.2,verbose=1)
```

## 9.Model Prediction

```python
# Example data with new 13 features

new_data = new_data = np.array([[0.02, 18, 2.5, 0, 0.45, 6.5, 65, 4.5, 1, 300, 15, 390, 5]])

#Applying scaling

new_data = scaler.fit_transform(new_data)

# Predict

prediction = model.predict(new_data)
print("Predicted House Price:",prediction)
```
