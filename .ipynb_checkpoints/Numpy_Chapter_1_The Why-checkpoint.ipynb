```python
import numpy as np
import time

size = 10000000

# Python lists
list1 = list(range(size))
list2 = list(range(size))

# NumPy arrays
arr1 = np.array(list1)
arr2 = np.array(list2)

# List timing
start = time.time()
result_list = [list1[i] + list2[i] for i in range(size)]
#print(result_list)
list_time = time.time() - start

# NumPy timing
start = time.time()
result_array = arr1 + arr2
#print("***************")
#print(result_array)
numpy_time = time.time() - start

print(f"List Time: {list_time:.5f} seconds")
print(f"NumPy Time: {numpy_time:.5f} seconds")
print(f"Speed Difference: {list_time / numpy_time:.2f}x faster")
```

### Purpose:
1. Handle large-scale calculations fast

### Why Not do the with same operations with other Data Structures like List?
1. Python lists → slow (loop-based)

### Problem Statement: We are going to build a simple brain… using only NumPy

### Problem:
Predict output using:

```python
y = 2x + 3
```

### Step 1: Import Necessary Libraries

```python
import numpy as np
```

### Step 2: Data Collection

```python
X = np.array([[1], [2], [3], [4]])  #X = Inputs
y = np.array([[5], [7], [9], [11]]) #y = actual outputs
```

### Step 3: Data Understanding

```python
X,y
```

```python
X.shape,y.shape
```

```python
X.ndim,y.ndim
```

### Step 4: Data Preparation, Model Building, Model Training

### Phase 1: The Core 4-Step SETUP STAGE

##### 4.1 Initialize weights randomly

```python
np.random.randn()
```

```python
w = np.random.randn(1,)
b = np.random.randn(1)

print("Initial w:", w)
print("Initial b:", b)
```

##### 4.2 Forward Pass (Prediction)

```python
def forward(X, w, b):
    return np.dot(X, w) + b

y_pred = forward(X, w, b)
print("Predictions:\n", y_pred)
```

##### 4.3 Loss Function (Error)

```python
def loss(y, y_pred):
    return np.mean((y - y_pred) ** 2)

print("Loss:", loss(y, y_pred))
```

##### 4.4 Backpropagation (Manual Gradient)

```python
def gradients(X, y, y_pred):
    n = len(X)
    
    dw = (-2/n) * np.dot(X.T, (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    
    return dw, db
```

### Phase 2: The Actual Training Phase

```python
print("Before Training:", forward(np.array([[5]]), w, b))
```

```python
learning_rate = 0.01

for i in range(1000):
    y_pred = forward(X, w, b)
    
    dw, db = gradients(X, y, y_pred)
    
    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    if i % 100 == 0:
        print(f"Epoch {i}, Loss: {loss(y, y_pred)}")
```

```python
print("After Training:", forward(np.array([[5]]), w, b))
```

### So... Learning = Updating weights and bias to reduce error.

### ==============================================================

## Step 5: Result

```python
print("Final w:", w)
print("Final b:", b)
```

##### Expected Result:
* w ≈ 2
* b ≈ 3
